## Outputs and modelling contract

The command above prints the final shape and class counts computed from the source files. `artifacts/data_profile.json` records raw shape, dtypes, missingness, duplicates, capped value counts, normalization, date parsing, vital thresholds and changes, unit inference, and integration counts. `artifacts/feature_schema.json` lists model features separately from target, metadata, and auxiliary columns.

The labelled CSV is written to `data/processed/triage_dataset.csv`. Unlabelled request identifiers are reported separately in `data/processed/unlabelled_intake_requests.csv`. Assessment columns other than the target are discarded. Metadata is retained for later patient-grouped, time-aware splitting.

Duration units are inferred as seconds; source metadata does not confirm them. Hours and days are retained for inspection; the modelling feature is log-transformed hours. Broad fixed vital bounds screen technical implausibility and are documented in the profile; they are not clinical thresholds. Unknown multi-value responses remain distinct from explicit none. Prior-request counts use only strictly earlier intake history, including unlabelled requests. Imputation and learned encoders belong in a train-only modelling pipeline after splitting.
